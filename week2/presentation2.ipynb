```python
import mlflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization, Input
from tensorflow.keras.callbacks import EarlyStopping

# 1. MLflow Experiment Set karein
mlflow.set_experiment("w2-syed-taha")
mlflow.autolog()

# ==========================================
# 1. BASELINE MLP
# ==========================================
with mlflow.start_run(run_name="mlp-baseline"):
    model_baseline = Sequential([
        Input(shape=(X_train.shape[1],)),
        Dense(64, activation='relu'),
        Dense(32, activation='relu'),
        Dense(1)
    ])
    model_baseline.compile(optimizer='adam', loss='mean_squared_error', metrics=['mae'])
    
    history_baseline = model_baseline.fit(
        X_train, y_train, 
        validation_split=0.2, 
        epochs=30, 
        batch_size=30,
        verbose=1
    )
    mlflow.log_param("model_type", "Baseline MLP")


# ==========================================
# 2. DEEPER DNN
# ==========================================
with mlflow.start_run(run_name="mlp-deeper"):
    model_deeper = Sequential([
        Input(shape=(X_train.shape[1],)),
        Dense(128, activation='relu'),
        Dense(64, activation='relu'),
        Dense(32, activation='relu'),
        Dense(16, activation='relu'),
        Dense(1)
    ])
    model_deeper.compile(optimizer='adam', loss='mean_squared_error', metrics=['mae'])
    
    history_deeper = model_deeper.fit(
        X_train, y_train, 
        validation_split=0.2, 
        epochs=50, 
        batch_size=50, 
        verbose=1
    )
    mlflow.log_param("model_type", "Deeper DNN")


# ==========================================
# 3. REGULARISED DNN
# ==========================================
with mlflow.start_run(run_name="mlp-regularised"):
    model_regularised = Sequential([
        Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
        BatchNormalization(),
        Dropout(0.1),
        Dense(32, activation='relu'),
        BatchNormalization(),
        Dropout(0.2),
        Dense(1)
    ])
    model_regularised.compile(optimizer='adam', loss='mean_squared_error', metrics=['mae'])
    
    early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
    
    history_reg = model_regularised.fit(
        X_train, y_train, 
        validation_split=0.2, 
        epochs=50, 
        batch_size=32, 
        callbacks=[early_stop],
        verbose=1
    )
    mlflow.log_param("model_type", "Regularised DNN")

print("All three models have successfully run and been logged in MLflow!")
```
